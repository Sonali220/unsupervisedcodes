```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Create data
data = {
    'Student': ['Amit', 'Priya', 'Rahul', 'Sneha', 'Karan', 'Neha'],
    'Math_Score': [85, 60, 90, 45, 70, 95],
    'Study_Hours': [6, 3, 7, 2, 4, 8]
}

df = pd.DataFrame(data)

# Select features
x = df[['Math_Score', 'Study_Hours']]

# Scaling
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)

# Apply K-Means
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(x_scaled)

# Display result
print(df)

# Plot clusters
plt.scatter(
    df['Math_Score'],
    df['Study_Hours'],
    c=df['Cluster']
)

plt.xlabel('Math Score')
plt.ylabel('Study Hours')
plt.title('K-Means Clustering')
plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import linkage,dendrogram

```

```python
x, _ = make_blobs(n_samples=30, centers=3, random_state=42)
labels = AgglomerativeClustering(n_clusters=3, linkage='single').fit_predict(x)
plt.plot()
dendrogram(linkage(x, method='single'))
plt.show()
```

```python
from sklearn.cluster import MiniBatchKMeans
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt

# Create dataset
x, _ = make_blobs(
    n_samples=1200,
    centers=5,
    n_features=5,
    random_state=42
)

# Create Mini-Batch K-Means model
mbk_means = MiniBatchKMeans(
    n_clusters=4,
    batch_size=95,
    random_state=42
)

# Train the model
mbk_means.fit(x)

# Plot clusters
plt.scatter(
    x[:, 0],
    x[:, 1],
    c=mbk_means.labels_,
    cmap='viridis'
)

# Plot cluster centers
plt.scatter(
    mbk_means.cluster_centers_[:, 0],
    mbk_means.cluster_centers_[:, 1],
    s=200,
    c='black',
    marker='x'
)

plt.title('Mini-Batch K-Means Clustering')
plt.show()
```
